```python
#split the data into training and test sets

from sklearn.model_selection import train_test_split

# Separate input features and target
X = df.drop(columns=["Class"])
y = df["Class"]

# Split before any resampling
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts().sort_index())

print("\nTesting target distribution:")
print(y_test.value_counts().sort_index())
```

Why split first? The test set must represent the naturally imbalanced data. We will apply SMOTE only to the training set, never to the test set.

1. What does stratify=y mean?

stratify=y tells train_test_split() to preserve approximately the
same class proportions in both the training and testing datasets as
in the original dataset.

In our project:
- y contains the target column, Class.
- Class = 0 means a legitimate transaction.
- Class = 1 means a fraudulent transaction.

2. What happens without stratify=y?

Imagine we randomly split 1,000 transactions:

Without stratification, the random split might put all 10 fraud cases into the training dataset, leaving no fraud examples in the test dataset.
Then we couldn't properly measure how well the model detects fraud.

The split attempts to preserve the original class proportions.

```python
from imblearn.over_sampling import SMOTE
```

```python
# Check the class distribution before balancing

print("Before SMOTE:")
print(y_train.value_counts().sort_index())

print("\nClass percentages:")
print(
    (y_train.value_counts(normalize=True).sort_index() * 100)
    .round(4)
)
```

```python
# Apply SMOTE

## SMOTE stands for Synthetic Minority Over-sampling Technique.
## It creates synthetic minority-class examples using nearby minority-class samples.

# Initialize SMOTE
smote = SMOTE(random_state=42)

# Resample training data only
X_train_balanced, y_train_balanced = smote.fit_resample(
    X_train,
    y_train
)

print("SMOTE completed successfully!")
```

```python
# Verify the transformation

print("Before balancing:")
print(y_train.value_counts().sort_index())

print("\nAfter balancing:")
print(y_train_balanced.value_counts().sort_index())

print("\nOriginal training shape:", X_train.shape)
print("Balanced training shape:", X_train_balanced.shape)
```

```python
# Visualize before and after

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

sns.countplot(x=y_train, ax=axes[0])
axes[0].set_title("Before SMOTE")
axes[0].set_xticks([0, 1], ["Legitimate", "Fraud"])
axes[0].set_xlabel("Class")
axes[0].set_ylabel("Number of Transactions")

sns.countplot(x=y_train_balanced, ax=axes[1])
axes[1].set_title("After SMOTE")
axes[1].set_xticks([0, 1], ["Legitimate", "Fraud"])
axes[1].set_xlabel("Class")
axes[1].set_ylabel("Number of Transactions")

plt.tight_layout()
plt.show()
```

a proper pipeline with scaling and SMOTE in the correct order. This

```python
# model comparison experiment

## Experiment A — Baseline: Train each classifier on the original imbalanced training data.

## Experiment B — SMOTE: Train the same classifiers
## using a pipeline that scales the data, applies SMOTE to the training folds, and trains the classifier.


# Build the model comparison pipeline

from sklearn.model_selection import train_test_split

# Create training and validation sets
X_fit, X_val, y_fit, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.20,
    stratify=y_train,
    random_state=42
)

print("Model training set:", X_fit.shape)
print("Validation set:", X_val.shape)
print("Final test set:", X_test.shape)

print("\nFraud cases:")
print("Training:", int(y_fit.sum()))
print("Validation:", int(y_val.sum()))
print("Final test:", int(y_test.sum()))


# We will use X_fit and y_fit for model training,
# X_val and y_val for model selection, and keep X_test and y_test untouched until final evaluation.
```

```python
# Define the classifiers

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier,
    HistGradientBoostingClassifier
)

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=1000, random_state=42
    ),
    "Decision Tree": DecisionTreeClassifier(
        random_state=42
    ),
    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        n_jobs=-1,
        random_state=42
    ),
    "Extra Trees": ExtraTreesClassifier(
        n_estimators=100,
        n_jobs=-1,
        random_state=42
    ),
    "Hist Gradient Boosting": HistGradientBoostingClassifier(
        random_state=42
    )
}
```

```python
# Train and compare both experiments

from sklearn.base import clone
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline as SklearnPipeline
from imblearn.pipeline import Pipeline as ImblearnPipeline
from imblearn.over_sampling import SMOTE

from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    average_precision_score,
    accuracy_score
)

results = []
fitted_models = {}

for model_name, model in models.items():

    for experiment in ["Baseline", "SMOTE"]:

        print(f"Training: {model_name} | {experiment}")

        if experiment == "Baseline":
            # Original imbalanced training data
            pipeline = SklearnPipeline([
                ("scaler", StandardScaler()),
                ("model", clone(model))
            ])

        else:
            # Scaling -> SMOTE -> classifier
            pipeline = ImblearnPipeline([
                ("scaler", StandardScaler()),
                ("smote", SMOTE(random_state=42)),
                ("model", clone(model))
            ])

        pipeline.fit(X_fit, y_fit)

        predictions = pipeline.predict(X_val)
        probabilities = pipeline.predict_proba(X_val)[:, 1]

        results.append({
            "Model": model_name,
            "Experiment": experiment,
            "Precision": precision_score(
                y_val, predictions, zero_division=0
            ),
            "Recall": recall_score(
                y_val, predictions, zero_division=0
            ),
            "F1": f1_score(
                y_val, predictions, zero_division=0
            ),
            "PR_AUC": average_precision_score(
                y_val, probabilities
            ),
            "Accuracy": accuracy_score(
                y_val, predictions
            )
        })

        fitted_models[(model_name, experiment)] = pipeline

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    ["Experiment", "PR_AUC"],
    ascending=[True, False]
).reset_index(drop=True)

display(results_df.round(4))
```

```python
## Select the best model for each experiment

best_models = (
    results_df.sort_values("PR_AUC", ascending=False)
    .groupby("Experiment", sort=False)
    .head(1)
)

display(
    best_models[
        ["Experiment", "Model", "Precision",
         "Recall", "F1", "PR_AUC", "Accuracy"]
    ].round(4)
)
```

```python
# Why the other models matter
# - Hist Gradient Boosting + SMOTE: Recall reached 82.28%, with precision of 62.50%.
# This is worth investigating if catching as many fraud cases as possible is the priority.
# - Logistic Regression + SMOTE: Recall reached 87.34%, but precision was only 5.42%.
# That means many legitimate transactions would be flagged.
# - Random Forest and Extra Trees: Both offer stronger precision-recall trade-offs in your current results.
```

```python
# Compare the leading candidates on the untouched test set

# Let's take three candidates forward:
# 1. Extra Trees — Baseline
# 2. Extra Trees — SMOTE
# 3. Hist Gradient Boosting — SMOTE

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    average_precision_score
)

final_candidates = {
    "Extra Trees - Baseline":
        fitted_models[("Extra Trees", "Baseline")],

    "Extra Trees - SMOTE":
        fitted_models[("Extra Trees", "SMOTE")],

    "Hist Gradient Boosting - SMOTE":
        fitted_models[("Hist Gradient Boosting", "SMOTE")]
}

test_results = []

for name, model in final_candidates.items():

    predictions = model.predict(X_test)
    probabilities = model.predict_proba(X_test)[:, 1]

    print(f"\n{'=' * 60}")
    print(name)
    print("=" * 60)

    print(classification_report(
        y_test,
        predictions,
        target_names=["Legitimate", "Fraud"],
        zero_division=0
    ))

    print("PR-AUC:", average_precision_score(
        y_test, probabilities
    ))

    test_results.append({
        "Model": name,
        "Precision": __import__("sklearn.metrics",
            fromlist=["precision_score"]).precision_score(
                y_test, predictions, zero_division=0
            ),
        "Recall": __import__("sklearn.metrics",
            fromlist=["recall_score"]).recall_score(
                y_test, predictions, zero_division=0
            ),
        "F1": __import__("sklearn.metrics",
            fromlist=["f1_score"]).f1_score(
                y_test, predictions, zero_division=0
            ),
        "PR_AUC": average_precision_score(
            y_test, probabilities
        )
    })

test_results_df = pd.DataFrame(test_results)
display(test_results_df.round(4))
```

```python
# Verify the confusion matrices

from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

for name, model in final_candidates.items():
    predictions = model.predict(X_test)
    cm = confusion_matrix(y_test, predictions)

    plt.figure(figsize=(5, 4))
    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=["Legitimate", "Fraud"],
        yticklabels=["Legitimate", "Fraud"]
    )
    plt.title(f"Confusion Matrix: {name}")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.tight_layout()
    plt.show()
```

```python
# Check the pipeline before deployment

best_model = final_candidates["Extra Trees - SMOTE"]

# Verify that the complete pipeline predicts directly
sample_predictions = best_model.predict(X_test.iloc[:10])
sample_probabilities = best_model.predict_proba(X_test.iloc[:10])[:, 1]

print("Sample predictions:", sample_predictions)
print("Fraud probabilities:", sample_probabilities.round(4))
print("Pipeline steps:", list(best_model.named_steps.keys()))
```

```python
# Final pipeline validation

from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    average_precision_score
)

# Verify the winning pipeline
best_model = final_candidates["Extra Trees - SMOTE"]

assert list(best_model.named_steps.keys()) == [
    "scaler", "smote", "model"
]

# Confirm the test data remains imbalanced
print("Test distribution:")
print(y_test.value_counts().sort_index())

# Re-evaluate the selected model
pred = best_model.predict(X_test)
proba = best_model.predict_proba(X_test)[:, 1]

print("\nFinal test metrics")
print("Precision:", round(precision_score(y_test, pred), 4))
print("Recall:", round(recall_score(y_test, pred), 4))
print("F1-score:", round(f1_score(y_test, pred), 4))
print("PR-AUC:", round(average_precision_score(y_test, proba), 4))

print("\nPipeline validation passed.")
```
